Importance Sampling

$\ \int_{x}^{} h(x) p(x) dx = \int_{x} \frac {h(x)p(x)}{g(x)} g(x) dx = \int_x h(x) w(x) g(x) dx$

Use 1 : Variance reduction.  
$h(x) = 10 \exp^{(-2|x-5|)}, x\sim U(0,10)$  
We want to calculate $E[h(x)]$. It can be calculated by sampling from $h(x)$, but this has high variance.  
$\int_{0}^{10} h(x) \frac{1}{10} dx = \int_{0}^{10} h(x) \frac{\frac{1}{10}}{N(5,1)} N(5,1) dx = \int_{0}^{10} h(x) w(x) N(5,1) dx$  
$N(5,1) = \frac{1}{\sqrt{2\pi}} \exp^{\frac{-(x-5)^2} {2}}$

```python
import numpy as np
np.random.seed(1)
```

```python
x = np.random.uniform(low=0,high=10,size=10000)
print('uniform rv',x.mean(), x.var())

print('If we directly estimate from h(x)')
h_x = 10*np.exp(-2*abs(x-5))
print('mean :{:.5f}, var:{:.5f}'.format(h_x.mean(), h_x.var()))
```

```python
def uniform_dist_mass(x, low=0, high=10):
  density_val = np.ones(x.shape)
  density_val *= 1/(high-low)
  density_val[np.where(x<low)[0]] = 0.0
  density_val[np.where(x>high)[0]] = 0.0
  return density_val
```

```python
from scipy.stats import norm
def normal_density_fn(x, mean=5.0, std = 1.0):
  return norm.pdf(x, loc=mean,scale=std)
```

```python
## Here we sample x from N(5,1)
x = np.random.normal(loc=5.0, scale=1.0, size=10000)
# print(x)
# uniform_dist_mass(x)
h_x = 10*np.exp(-2*abs(x-5))
w_x = (1/10)/normal_density_fn(x)
y = h_x * w_x
print(y.mean(), y.var())
```
