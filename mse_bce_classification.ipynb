```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate sample values from -20 to 100
x = np.linspace(-20, 10, 100) # w^T x
# Define the first function: x^2
y1 = (x-1)**2 # true label as 1

# Define the second function: -log(sigmoid(x))
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

y2 = -np.log(sigmoid(x))

# Create the plot
plt.figure(figsize=(10, 6))

# Plot x^2
plt.plot(x, y1, label='mse_loss = $(x-1)^2$', color='blue')

# Plot -log(sigmoid(x))
plt.plot(x, y2, label='BCE loss: $-\log(\sigma(x))$', color='orange')

# Add labels, legend, and title
plt.xlabel('x')
plt.ylabel('y')
plt.title('BCE and MSE loss')
plt.legend()
plt.grid()
# Show the plot
plt.show()
```

$\frac{\partial(log(\sigma(x)))}{\partial x} = \sigma(-x)$ 

```python
# Generate sample values from -20 to 100
x = np.linspace(-20, 10, 100) # w^T x
# Define the first function: x^2
grad_y1 = 2*(x-1) # true label as 1

# Define the second function: -log(sigmoid(x))
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

grad_y2 = sigmoid(-x)

# Create the plot
plt.figure(figsize=(10, 6))

# Plot x^2
plt.plot(x, grad_y1, label='grad_mse_loss = $(x-1)*2$', color='blue')

# Plot -log(sigmoid(x))
plt.plot(x, grad_y2, label='BCE loss: $\sigma(-x)$', color='orange')

# Add labels, legend, and title
plt.xlabel('x')
plt.ylabel('gradient')
plt.title('BCE and MSE loss gradients')
plt.legend()
plt.grid()
# Show the plot
plt.show()
```
