$KL(Q || P) = \sum_x q(x) log \frac{q(x)}{p(x)}$  
In practice, this is estimated as avg of $-log r, where r=\frac{p(x)}{q(x)}$
* this is unbiased estimate, but has high variance
* It can be negative at some samples x, but the KL>0

Another estimate
* Avg of r-1-log r
* This is unbiased
* And always >0



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_function():
    """Plots the function y = x - 1 - log(x) for x in the range (0, 10)."""

    # Generate 200 evenly spaced x values between 0.1 and 10 (avoiding x=0 due to log)
    x = np.linspace(0.1, 10, 400)  # Changed start to 0.1 to avoid log(0)

    # Calculate the corresponding y values
    y = x - 1 - np.log(x)

    # Create the plot
    plt.plot(x, y)

    # Add labels and title
    plt.xlabel("x")
    plt.ylabel("y = x - 1 - log(x)")
    plt.title("Plot of y = x - 1 - log(x)")

    # Add a grid for better readability (optional)
    plt.grid(True)

    # Show the plot
    plt.show()


# Call the function to generate and display the plot
plot_function()
```
